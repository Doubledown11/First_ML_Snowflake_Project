```
import numpy 

def correlation_ratio(categories, values):
    """
    correlation_ratio:
        This function takes in a categorical feature and a a numeric feature and
        it returns a measure of how much of the variance in values is explained 
        by categories.
    """

    # Converts categories into pandas categorical object. 
    categories = panda.Categorical(categories)
    # Extracts the unique category levels. 
    cat_levels = categories.categories
    # Compuates the overall mean of the given numeric category saved in the variable values.
    y_mean = values.mean()

    # Below builds the between-group sum of squares:
    # For each category cat in cat_levels: 
    # values[categories==cat] shows all numeric values belonging in that category. 
    numerator = sum([
        # len(... is the number of observations in that category.
        len(values[categories == cat]) *
        # Below si the squared difference between that category's mean and the overall mean. 
        (values[categories == cat].mean() - y_mean) ** 2
        for cat in cat_levels
    ])
    # Sum above adds the quantity calculated within over all categories.

    # Below is total sum of squares / AKA the total variance in the numeric variable:
    denominator = sum((values - y_mean) ** 2)
    
    # below cpmputes and returns the correlation ratio.
    # If denominator is not equal to 0, compute the proportion of variance explained by categories.
    # If denominator is equal to zero, all values are identical (no variance) and thus we return 0.
    return numpy.sqrt(numerator / denominator) if denominator != 0 else 0


def mixed_correlation(df):
    """
    mixed_correlation:
        Takes in a pandas dataframe and returns a correlation matrix that 
        works for numeric, categorial, and mixed pairs.
    """
    
    cols = df.columns # Stores columns in the dataframe.
    # Creates an empty square dataframe to store the results. 
    # rows/index as well as the columns of this new dataframe 'corr' are set 
    # to the columns of the 'cols' dataframe. All values are also set to floats.
    corr = panda.DataFrame(index = cols, columns = cols, dtype = float) 

    # Below loops over every pair of columns. 
    # This ensures a full matrix.
    for col1 in cols:
        for col2 in cols:
            # If the 2 columns are the same.
            if col1 == col2:
                # Set the correlation to 1, and skip the rest of the loop.
                corr.loc[col1, col2] = 1.0
                continue

            # Below extracts the 2 columns being compared.
            x = df[col1]
            y = df[col2]

            # Numeric ↔ Numeric
            # If both columns are numeric, we use a Pearson correlation, and store the results in the matrix.
            if panda.api.types.is_numeric_dtype(x) and panda.api.types.is_numeric_dtype(y):
                corr.loc[col1, col2] = x.corr(y)

            # Categorical ↔ Categorical
            # If both columns are non-numeric, we use a Cramer's V correlation, and store the results in the matrix.
            elif x.dtype == "object" and y.dtype == "object":
                corr.loc[col1, col2] = cramers_v(x, y)

            # Numeric ↔ Categorical
            # If one category is numeric and the other is non-numeric, this checks which is 
            # numeric and orders them correctly for the n_r
            else:
                if panda.api.types.is_numeric_dtype(x):
                    corr.loc[col1, col2] = correlation_ratio(y, x)
                else:
                    corr.loc[col1, col2] = correlation_ratio(x, y)

    return corr # Return the correlation matrix.
```

```
# Method 2: A Streamlit Interactive App:
"""
Cramér’s V is a correlation measure for 
categorical variables. 

This measure us used to describe the relationship
between two non-numeric features, such as: 
    fur_color 
    facial_profile
    paw_pad_texture
    species 
    Or other such as
        brand, category, type...

Possible return values from a cramers analysis:
    0.0 -- No association at all. 
    0.1-0.3 -- Weak association.
    0.3-0.6 -- Moderate association. 
    0.6-1.0 -- Strong association.

How does it work?
    1) Build a contingency table: 
        Which is like a pivot table 
        counting combinations of categories 

    2) Run a chi-square test on that table 
        Measures how different the observed counts
        are from random chance. 

    3) Normalize the chi-square statistic 
        This allows the result to always be beween 
        0 and 1.

This us a useful meausre since categorical features 
don't have numeric distance, thus you can't compute 
a Pearson correlation. 

"""

from scipy.stats import chi2_contingency


def cramers_v(x, y):
    """
    Calculates cramer's V given the 2 input variables, x and y. This is a 
    measure of how closely related the 2 variables are. 

    x and y are two different columns from the dataframe being processed 
    in the function categorical_correlation. 
    """
    
    confusion_matrix = panda.crosstab(x, y) # Create contingency table.

    # If either variable has only one category → no association
    if confusion_matrix.shape[0] == 1 or confusion_matrix.shape[1] == 1:
        return 0.0

    chi2 = chi2_contingency(confusion_matrix)[0] # runs chi square test and get chi square value.
    n = confusion_matrix.sum().sum() # holds number of observations.
    phi2 = chi2 / n # calculates chi-square value.
    r, k = confusion_matrix.shape # r is number of categroeis in x, and k is number in y.

    # Bias correction
    phi2corr = max(0, phi2 - ((k - 1) * (r - 1)) / (n - 1)) # correct chi-square for small sample size 
    rcorr = r - ((r - 1) ** 2) / (n - 1) # Corrects effective number of categories
    kcorr = k - ((k - 1) ** 2) / (n - 1) # Corrects effective number of categories

    # Below calculates the denominatorand then checks if it is invlaid. 
    denom = min((kcorr - 1), (rcorr - 1))
    if denom <= 0:
        return 0.0 # If denominator is invalid, return 0.

    return numpy.sqrt(phi2corr / denom) # Retuens the cramers V value between 0 and 1. 

# Below creates a full correlation matrix for categorical features only.
def categorical_correlation(df):
    """
    Takes a dataframe as input and computes Cramer's V for every pair of columns.
    """
    
    cols = df.columns # Holds a list of the dataframes column names.
    corr = panda.DataFrame(index = cols, columns = cols, dtype = float) # Creates an empty square matrix to fill with Cramers V values.

    # Below loops over every pair of values between column 1 and column 2.
    for col1 in cols:
        for col2 in cols:
            # If column 1 and column 2 are equal (In the diagonal), the feature is said to be perfectly correlated with itself. 
            if col1 == col2:
                corr.loc[col1, col2] = 1.0
            else:
                corr.loc[col1, col2] = cramers_v(df[col1], df[col2])

    return corr # Returns the correlation matrix.



# Below creates a header for the streamlit application.
streamlit.header("🐻 Bear Feature Relationships (Categorical)")

# Below creates the columns for the selectors
col1, col2 = streamlit.columns(2)
EXCLUDE = {"id", "bear_id", "id_x", "id_y"} # Defines the columns to be ignored.

# Below filters out the id column from the list of valid features in the dropdown list. 
valid_features1 = [c for c in df_raw.columns if c.lower() not in EXCLUDE]
valid_features2 = [c for c in df_raw.columns if c.lower() not in EXCLUDE]

# Creates 2 dropdowns so the user can choose: 
x_feature = col1.selectbox("Select X-axis Feature", valid_features1)
y_feature = col2.selectbox("Select Y-axis Feature", valid_features2)

# If the user selects the same feature twise, display a warning.
if x_feature == y_feature:
    streamlit.warning("Please select two different features.")

# If 2 different features are selected, we build the heatmap/matrix.
else:
    # Below code builds the heatmp with the users' desired features:
    heatmap_data = (
        df_raw
        .groupby([x_feature, y_feature])
        .size()
        .rename("count")
        .reset_index()
    )
    
    # Below renders the heatmap using Altair:
    chart = (
        altair.Chart(heatmap_data)
        .mark_rect()
        .encode(
            x = altair.X(f"{x_feature}:N", title = x_feature),
             = altair.Y(f"{y_feature}:N", title = y_feature),
            color = altair.Color("count:Q", scale=altair.Scale(scheme = "browns")),
            tooltip = [x_feature, y_feature, "count"]
        )
        .properties(
            width = 400,
            height = 400,
            title = f"Heatmap: {x_feature} vs {y_feature}"
        )
    )
    
    streamlit.altair_chart(chart, use_container_width = True)
```

```
# Feature Scaling: 
# This is a data preprocessing technique used to standardize the range of 
# indenpendent variables (AKA features) of the data sample. This helps to ensure 
# that features with larger value ranges do not disporportionately affect the model's 
# learning process.

# Process of this cell:
# ID numerical vs categorical features
# Scale numerical features
# Encode categorical features 
# Recombine everything into a clean, final dataframe.


# We use scikit-learn to do this with 'mean centering' (mean = 0) unit variance (SD=1) 
# This sets all variables' means and SD to 0 and 1, respectively. 

# StandardScaler scales numeric features.
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.compose import ColumnTransformer
# OneHotEncoder converts categorical features into binary columns.   
from sklearn.preprocessing import OneHotEncoder

# Identify numerical and categorical columns
# This is accomplished by looking into the data  types of each column in X_train 
# and then split them into 2, Numerical (integers and floats) and categorical features (Strings) 
numerical_features = X_train.select_dtypes(include = ['int64', 'float64']).columns
categorical_features = X_train.select_dtypes(include = ['object']).columns


# Below prints which columns belong to which group.
print("Numerical features:", numerical_features.tolist())
print("Categorical features:", categorical_features.tolist())

# Below StandardScaler is used to scale numerical features. 
scaler = StandardScaler()
# For each numerical columns the scaled value is calculated as = (x-mean)/standard deviation 
# Below ysed fit_transform() on the training data as this learns the meand standard deviation.
X_train_scaled_num = scaler.fit_transform(X_train[numerical_features])

# Below uses .transform() on the test data.
X_test_scaled_num = scaler.transform(X_test[numerical_features])


# Below uses OneHotEncoder to convert categorical features using one-hot encoding
# handle_unknown='ignore' is used if the test set contains a category not seen in 
# training, it avoids errors by ignoring it.
# sparse_output=False returns a dense numpy array instead of a sparse matrix.
# OneHot encoding is a way to convert categorical test lables into numerical columns so ML models can use them.
onehot = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
X_train_scaled_cat = onehot.fit_transform(X_train[categorical_features])
X_test_scaled_cat = onehot.transform(X_test[categorical_features])

# Below gets the new feature names after onehotencoder is done above.
cat_feature_names = onehot.get_feature_names_out(categorical_features)

# Combine numerical and categorical features
# hstack() means horizontal stack, which is a merge of 2 arrays side by side into one final 
# feature matrix.
X_train_scaled = numpy.hstack([X_train_scaled_num, X_train_scaled_cat])
X_test_scaled = numpy.hstack([X_test_scaled_num, X_test_scaled_cat])

# Convert to DataFrame with proper column names
# Below restores column names, keeps the original row index, and makes the data easier to inspect and use in models.
all_feature_names = list(numerical_features) + list(cat_feature_names)
X_train_scaled = panda.DataFrame(X_train_scaled, columns=all_feature_names, index=X_train.index)
X_test_scaled = panda.DataFrame(X_test_scaled, columns=all_feature_names, index=X_test.index)

print("\n✅ Features scaling completed!")
print('-' * 35) 

print("\n📊 Scaled Data Dimension:")
print(f"Scaled training features shape: {X_train_scaled.shape}")
print(f"Scaled testing features shape: {X_test_scaled.shape}")
print('-' * 35) 

# Show scaling effect for numerical features
# This shows that demonstrates that scaling worked.
if len(numerical_features) > 0:
    first_num_feature = numerical_features[0]
    print("\n📊 Scaling Effect (first numerical feature):")
    print(f"Original {first_num_feature}: mean={X_train[first_num_feature].mean():.3f}, std={X_train[first_num_feature].std():.3f}")
    print(f"Scaled {first_num_feature}: mean={X_train_scaled[first_num_feature].mean():.3f}, std={X_train_scaled[first_num_feature].std():.3f}")
print('-' * 35)
```

# Section 3 - Part 2: Machine Learning Model Training

```
# This cell does the following:
# Trains a logistic regression classifier
# Makes predictions 
# Evaluates the mode using multiple metrics 
# Prints a readable summary

# Logistic Regression: 
# A statistical method used for binary classification tasks, where the goal 
# is to predict 1 of 2 possible outcomes based on input features. It works by 
# estimating the probability that a given input belongs to a particular class 
# using the logistic (Sigmoid) function. This function maps any real number into a 
# value between 0 and 1. This probability is then used to classify input by 
# applying a threshold, commonly 0.5. 

# Import logistic model and classification metrics
# LogisticRegression is used for the machine learning model we are training.
from sklearn.linear_model import LogisticRegression

# accuracy_score used to get the % of correct predictions.
# classification_report used to get precision, recall, and F1 score (measures how well the model balances precision and recall). 
# matthews_corrcoef used to get a balanced metric for binary classification.
from sklearn.metrics import accuracy_score, classification_report, matthews_corrcoef


# Logistic Regression using scikit-learn
print("🔧 Training Logistic Regression Model...")

# Below creates a logistic regression model. 
log_reg_model = LogisticRegression(random_state = 42) # random_state is used for reproducibility
# .fit() on the logression model trains the model using the scaled features and the target labels.
log_reg_model.fit(X_train_scaled, y_train)

# Make predictions (outputs class labels directly)
# Predicts class labels of 0 or 1 for: the training and test sets. 
# These predictions are used to evalute performance.
log_reg_train_pred = log_reg_model.predict(X_train_scaled)
log_reg_test_pred = log_reg_model.predict(X_test_scaled)

# Compute evaluation metrics
# Below calculates the accuracy score (% correct predictions) for both the training and test sets.
logreg_train_acc = accuracy_score(y_train, log_reg_train_pred)
logreg_test_acc = accuracy_score(y_test, log_reg_test_pred)
# Below calculates the Mathews Correlation Coefficient for both the training and test sets.
# Mathews Correlation Coefficient can returns calues of:
# +1 --> Perfect prediction 
# 0 --> Random guesing 
# -1 --> Total disagreeemnt
# MCC is more reliable than accuracy when classes are imbalanced. This is because MCC 
# evaluates all 4 parts of the confusion matrix. (True positives, True negativees, False positives, false negatives)
logreg_train_mcc = matthews_corrcoef(y_train, log_reg_train_pred)
logreg_test_mcc = matthews_corrcoef(y_test, log_reg_test_pred)

# Below generates a report on the model's performance on the test set. 
# This models is generates with the input of y_test which are the true labels and teh log_reg_test_pred which
# are the model's predicted lables.
test_class_report = classification_report(y_test, log_reg_test_pred)

print("✅ Logistic Regression model trained!")
print('-' * 35)

print(f"📊 Logistic Regression Results:")
print(f"   Training Accuracy: {logreg_train_acc:.4f}")
print(f"   Testing Accuracy:  {logreg_test_acc:.4f}")
print(f"   Training MCC:      {logreg_train_mcc:.4f}")
print(f"   Testing MCC:       {logreg_test_mcc:.4f}")
print('-' * 35)

print("\nClassification Report (Test Set):")
print(test_class_report)
print('-' * 35)
```

```
# This cell does the following:
# Trains a random forect model
# Makes predictions on training and test data 
# Computes accuracy and MCC 
# Generates a classfication report 
# Prints a clean summary of performance.

# Random Forest Classifier:
# This is an esemble ML algorithm used for classification and regression tasks. 
# It builds multiple decision trees during training and meges their results to improve 
# accuracy and control overfiting.
# Each decision tree is trained on a random subset of the data, which introudces 
# diversity among the trees. For classification, the final prediction is made by majority voting among all the trees,
# while for regression is is the average of the trees' predictions. 


# Import ensemble methods and detailed metrics
# RandomForestClassifier is used for the random forest machine learning model. 
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, matthews_corrcoef, classification_report

# Random Forest using scikit-learn
print("🌲 Training Random Forest Classifier...")
print('-' * 35)

# Below initializes and trains the random forest model. 
# n_jobs =- 1 ensures all CPU cores are used for faster training. 
# .fit() trains the forest on the scaleed training data.
rf_model = RandomForestClassifier(random_state = 42, n_jobs =- 1) # Use parallel processing
rf_model.fit(X_train_scaled, y_train)

# How does the random forest work? 
# Each tree sees a random subset of rows 
# Each tree sees a random subset of features
# Trees learn differernt patterns 
# Final predictoin is made from a majority vote across all trees. 
# And this reduces overfitting and increases robustness 
# (robustness means the model can handle a wider range of input data) 

# Make predictions
# Below predicts class lebals for training and test sets.
# these predictions are used to evaluate performance.
rf_train_pred = rf_model.predict(X_train_scaled)
rf_test_pred = rf_model.predict(X_test_scaled)

# Calculate metrics for model performance evaluation.  
rf_train_acc = accuracy_score(y_train, rf_train_pred)
rf_test_acc = accuracy_score(y_test, rf_test_pred)
rf_train_mcc = matthews_corrcoef(y_train, rf_train_pred)
rf_test_mcc = matthews_corrcoef(y_test, rf_test_pred)
test_class_report = classification_report(y_test, rf_test_pred)


# Print the results of the models performance.
print("✅ Random Forest model trained!")
print('-' * 35)

print(f"📊 Random Forest Results:")
print(f"   Training Accuracy: {rf_train_acc:.4f}")
print(f"   Testing Accuracy:  {rf_test_acc:.4f}")
print(f"   Training MCC:      {rf_train_mcc:.4f}")
print(f"   Testing MCC:       {rf_test_mcc:.4f}")
print("\nClassification Report (Test Set):")
print(test_class_report)
print('-' * 35)
```

```
# This cell does the following: 
# Trains a support vector machine model 
# Makes predictions on the training and test data sets 
# Calculate s the accuracy and MCC of the model
# Generates a classfiication report 
# Prints a summary of the model's performance.

# Support Vector Machines:
# These are supervised learning models that are used for classification and regression. 
# They work by finding the optimal hyperplane tha tbest separates data points of different 
# classes with the maximum margin. 
# SVMs focus on the data points closest to the decision boundary, called support vectors, which 
# determine the position and oreintation of the hyperplane. 
# SVM can make data that is unable to be linearly separated into data that can be linearly 
# separated using kernel functions to tramsform the data into a higher dimensional space where
# a linear separation is possible.
# SVMs are effective in high dimensional spaces and are robust to overfitting, especially in 
# cases where the number of features exceeds the number of samples. 

# Import SVM and detailed metrics
# SVC is used as the support vector machine model. 
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, matthews_corrcoef, classification_report

# Support Vector Machine using scikit-learn
print("🤖 Training Support Vector Machine...")
print('-' * 35)

# Below initializes ans trains the SVM
svm_model = SVC(random_state = 42)
svm_model.fit(X_train_scaled, y_train)

# Make predictions on the class lables for the training and test sets. 
# These lables predictions are used to evluate performance.
svm_train_pred = svm_model.predict(X_train_scaled)
svm_test_pred = svm_model.predict(X_test_scaled)

# Calculate evaluation  metrics
svm_train_acc = accuracy_score(y_train, svm_train_pred)
svm_test_acc = accuracy_score(y_test, svm_test_pred)
svm_train_mcc = matthews_corrcoef(y_train, svm_train_pred)
svm_test_mcc = matthews_corrcoef(y_test, svm_test_pred)
test_class_report = classification_report(y_test, svm_test_pred)


# Print the results:
print("✅ SVM model trained!")
print('-' * 35)

print(f"📊 SVM Results:")
print(f"   Training Accuracy: {svm_train_acc:.4f}")
print(f"   Testing Accuracy:  {svm_test_acc:.4f}")
print(f"   Training MCC:      {svm_train_mcc:.4f}")
print(f"   Testing MCC:       {svm_test_mcc:.4f}")
print("\nClassification Report (Test Set):")
print(test_class_report)
print('-' * 35)
```

# Section 3 - Part 3: Benchmarking of Machine Learning Algorithms

Benchmarking means we are comparing various ML models to see which perform the best
and/or are best for the specific use case. 

We wish to seelct a model that can generalize well on new, unseen data and one that can provide actionable insights. 

We are careful to be aware of the following 2 points:
1) Model Overfitting: We can evaluate how well the model generalizes on new data by 
testing the degree the algorithm overfits the data. 
2) Model Interpretability: We ensure our model yields actional insight by understanding the underlying reasons as to 'why' the model made the predictions it does. We can do this by tracing how input features are influencing the output. 

```
# Assessing Overfitting:
# Overfitting is the phenomenon of when a model performs better on the data its input
# it was trained on rather than new data. 
# This says that the model memorized its inputs, rather then learning a general pattern.
# The formula of overfitting degree = TrainingPerformance - TestingPerformance is used to 
# assess overfitting. A larger difference means the model is more overfitted.

# Configure Altair for interactive visualizations
# enables a json transformer. 
# Uses the opaque theme, whcih gives charts a consistent visual style.
altair.data_transformers.enable('json')
altair.theme.enable('opaque')

# Below builds a dataframe that compares all models' accuracies.
# In this dataframe each row corresponds to a model, and each column contains data on the model
# Column1 is the model name, column2 is the training accuracy and column3 is the testing accuracy.
model_acc = panda.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest', 'SVM'],
    'Training_Accuracy': [logreg_train_acc, rf_train_acc, svm_train_acc],
    'Testing_Accuracy': [logreg_test_acc, rf_test_acc, svm_test_acc]
})

# Below calculates the overfitting score as the training accuracy minus the testing accuracy.  
# A small difference means the model generalizes well, while a large difference means the model is overfitting.
model_acc['Overfitting'] = model_acc['Training_Accuracy'] - model_acc['Testing_Accuracy']

# Below builds a dataframe as above, but instead compares the  models' MCC avlues. 
model_mcc = panda.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest', 'SVM'],
    'Training_MCC': [logreg_train_mcc, rf_train_mcc, svm_train_mcc],
    'Testing_MCC': [logreg_test_mcc, rf_test_mcc, svm_test_mcc]
})

# Below calculates the overfitting score as above, but does it for the MCC score. 
model_mcc['Overfitting'] = model_mcc['Training_MCC'] - model_mcc['Testing_MCC']

# Print the results.
print("📊 Model Comparison:")
print('-' * 35)

print("Accuracy:")
print(model_acc.round(4)) # Rounds the numbers to 4 decimal places.
print('-' * 35)

print("MCC:")
print(model_mcc.round(4))
```

```
# Model Interpretability: 
# Interpretable ML models are those that provide the variable coefficients that
# directly dictate the relative degree at which it influence the target y values. 
# In linear models this can be summarized using y = m1x1 +m2m2 +...+b 
# y = dependent variable 
# mn = variable coefficients 
# xn = the independent variables 
# b = the baseline value.

# mn coefficients are direct measures of their influence on the prediction of y.
# Where larger values absolute coefficient valu means that it has a strongre impact on the prediction of y.

# This cell does the following: 
# Extracts the learned coefficients from the logistic regression model
# Pairs each coefficient with its corresponding feature name
# Computes the aboslute value of each coefficient 
# Sorts features by absolute values 
# Prints the top 5 most influential features. 

# Intrepreting Logistic Regression Models:

# Below extracts the coefficients from the trained loigstic regression model
# Logistic regression learns one coefficient per feature. These coefficients 
# represent how strongly each feature pushes the predictoin toward class 1 or 0. 
# coef_ is a 2D array and we select from index 0 in coef_ 
# This gives a numpy array in coefficients.
coefficients = log_reg_model.coef_[0]


# Below creates a DataFrame that links the transformed feature names to coefficients: 
# all_feature_names contains: numerical feature names, one hot encoded categorical feature names
# Then each feature is paired with its coefficient.
logreg_feature_importance = panda.DataFrame({
    'feature': all_feature_names,  # Using all_feature_names from the py_feature_scaling cell
    'coefficient': coefficients
})

# Below calculates the absolute value of the coefficients to use as 'importance'
# Coefficients can be positive or negative 
# But note importance is a measure of magnitide, so -1.5 is as strong as +1.5 
# Thus absolute value is used to measure strength. 
logreg_feature_importance['abs_coefficient'] = numpy.abs(logreg_feature_importance['coefficient'])

# Below sorts the features by importance in descending order 
# Large positive coefficient -> increase probability of class 1 
# Large negative coefficient --> Decreases the probability of class 2 
# Small coefficient Means the feature has little influence.
logreg_feature_importance = logreg_feature_importance.sort_values('abs_coefficient', ascending = False)

# Print the results
print("✨ Top 5 Most Important Features (Logistic Regression):")
print(logreg_feature_importance[['feature', 'coefficient', 'abs_coefficient']].head())

```

```
# Below splits the dataset into training and testing sets:
X_train, X_test, y_train, y_test = train_test_split(
    X, y, # x is the features and y is the target labels 
    test_size = 0.2, # Sets the test set to be 20% of the data 
    random_state = 42, # Random_state is a seed for the random number generator.
    stratify = y # This ensures that both the training and testing sets have the szame class distirbution az the original dataset.
)

streamlit.write("✅ Data preparation completed!")

# Below displays the spliut summary, and this shows:
# Number of training samples, number of testing samples, percentage split, number of features (Columns)
streamlit.subheader("📊 Data Split Summary:")
streamlit.write(f"Training set: `{X_train.shape[0]} samples ({X_train.shape[0]/len(X)*100:.1f}%)`")
streamlit.write(f"Testing set: `{X_test.shape[0]} samples ({X_test.shape[0]/len(X)*100:.1f}%)`")
streamlit.write(f"Number of features: {X_train.shape[1]}")

# Below shows the class distribution in each split:
# This is done to ensure stratification worked, confirms no class was lost during splitting,
# helps detect imbalance issues early. 
streamlit.subheader("🎯 Class Distribution:")
streamlit.write(f"Training set: `{y_train.value_counts().sort_index().to_dict()}`")
streamlit.write(f"Testing set: `{y_test.value_counts().sort_index().to_dict()}`")
```

```
# We perform the following preprocessing: 
# 1) StandardScaler for numerical features. This transforms features to have mean=0 
# and std=1 
# 2) OneHotEncoder for categorical features (fur_color, facial_profile, paw_pad_texture)
# This converts categorical variables into binary columns. 

# Feature scaling and preprocessing
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Identify numerical and categorical columns by looking at the data types in X_train
# Selects int64 and float64 as the numerical features and object s as the categorical features.
numerical_features = X_train.select_dtypes(include = ['int64', 'float64']).columns
categorical_features = X_train.select_dtypes(include = ['object']).columns

# Below scales the numerical features
# StandardScaler scales the features so mean equals zero and standard deviation is equal to
# 1. This is needed for the logistic rergression, SVM, and KNN models.
# This is done to prevent large scale features (big values like weight) from dominating 
# small scale ones (Like age). 
scaler = StandardScaler()
X_train_scaled_num = scaler.fit_transform(X_train[numerical_features])
X_test_scaled_num = scaler.transform(X_test[numerical_features])

# Below handles categorical features by converting each into a binary column.
# We also ensure to ignore any categories not seen inthe test set.
onehot = OneHotEncoder(sparse_output = False, handle_unknown='ignore')
X_train_scaled_cat = onehot.fit_transform(X_train[categorical_features])
X_test_scaled_cat = onehot.transform(X_test[categorical_features])

# Get feature names after one-hot encoding and replace spaces with underscores
cat_feature_names = [] # Holds the names of the categorical features.
# Below creates feature names that are to be added to the above list.
# zip() takes 2 or more lists and pairs their ele
for feature, categories in zip(categorical_features, onehot.categories_):
    for category in categories:
        cat_feature_names.append(f"{feature}_{category}".replace(' ', '_').lower())

# Below combines numerical and categorical features matrices into a single one.
X_train_scaled = numpy.hstack([X_train_scaled_num, X_train_scaled_cat])
X_test_scaled = numpy.hstack([X_test_scaled_num, X_test_scaled_cat])

# Below converts a list of all the feautre names to a DataFrame with proper column names
# This is done for easier dubgging, easier feature importance analysis, easier visuialization,
# and it preserves row alignment with the original dataset.  
all_feature_names = list(numerical_features) + cat_feature_names
X_train_scaled = panda.DataFrame(X_train_scaled, columns = all_feature_names, index = X _train.index)
X_test_scaled = panda.DataFrame(X_test_scaled, columns = all_feature_names, index = X_test.index)


# Below displasy the results using Streamlit:
streamlit.write("Feature scaling completed!")

streamlit.subheader("Feature Processing:")
streamlit.write(f"Numerical features: `{numerical_features.tolist()}`")
streamlit.write(f"Categorical features: `{categorical_features.tolist()}`")

streamlit.subheader("Scaled Data Dimensions:")
streamlit.write(f"Training features: `{X_train_scaled.shape}`")
streamlit.write(f"Testing features: `{X_test_scaled.shape}`")

# Display first few encoded feature names
streamlit.write("\nFirst few encoded feature names:")
streamlit.write(all_feature_names[:10])
```

```
# This cell encodes the target variable (Bear Species) to numerical values using scikit-learn's 
# LabelEncoder. This is done since ML models cannot work with text labels such as a bears species name
# thus by converting the species names into numerical value, without implying any mathematical 
# relationship between them, the ML model can work with the data. 

# Each unique bear species will be assigned a unique integer value, 
# The encoding preserves the categorical nature of the species while making
# it suitable for model training.
# LableEncoder is scitkit-learn's categorical lable encoder.
from sklearn.preprocessing import LabelEncoder

# Below encodes the target variable
# Below initializes an encoder object that wil learn the mapping from species to integers.
label_encoder = LabelEncoder() 
# Below fit_transform() is used to learn all unique species in the traingin set, 
# and to convert each species into an integer.
# We fit on traingin data only to avoid data leakage.
# What does it mean to fit? Learn patterns or parameters from the training data.
y_train_encoded = label_encoder.fit_transform(y_train)

# Below transforms the test lables using the same mapping as the training set did.
y_test_encoded = label_encoder.transform(y_test)
```

Section 4 - Part 3: Experiment Tracking Setup

```
# We do experiment tracking using snowflakes ML experiemnt tracking 
# This allows us to log and compare different models and their hyperparameters.

# Below imports snowflake's experiment tracking libary. This lets us log models, metrics 
# hyperparameters, compare runs, store artifacts, and keep a history of experiments.
from snowflake.ml.experiment.experiment_tracking import ExperimentTracking

# Create an ExperimentTracking object that is tracking the current snowflake session.
exp = ExperimentTracking(session = sesh)

# Below sets the experiment name
experiment_name = "Bear_Classification_Experiment"
# Below ensures snowflake logs all future model runs. 
exp.set_experiment(experiment_name)

# Below confirms the ExperimentTracking is initialized.
st.write(f"✅ Experiment Tracking Initialized: `{experiment_name}`")
```

```
# This cell does the following:
# Trains the random forest model, evaluates the model, logs hyperparameters to snowflake,
# logs metrics, and stores the results inside of the ExperimentTracker

# Below imports what we need for the model and its evaluation.
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, matthews_corrcoef
from datetime import datetime

# Define Hyperparameters. These are the settings that control how the random froiest behaves.
params = {
    # Below builds 100 decision trees.
    "n_estimators": 100,
    # Below ensures trees can only be 3 levels deep (prevents overfitting)
    "max_depth": 3,
    # Below ensures each leaf must have at least 5 samples.
    "min_samples_leaf": 5,
    # Below ensures each split considers a sqrt(num_features) number of random features.  
    'max_features': 'sqrt',
    # Below ensures reproducibility.
    "random_state": 42
}

# Below creates a unique run name with the timestamp. This ensures that every 
# experiments get a unique ID, and thus makes it easier to compare runs later one.
run_name = f"bear_baseline_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

# Below starts an experiment run:
# Everything inside the with block is contained to this experiment run.
# Inside this block we log, hyperparameters, metrics, and model performance.
with exp.start_run(run_name):
    # Below logs hyperparameters. IE) Stores the models' setting inside snowflake.
    exp.log_params(params)

    # Below trains the model by fiurst rceating a random firest classifier using the 
    # hyperparameters. Then .fit() trains it on the scaled training data.  
    model = RandomForestClassifier(**params)
    model.fit(X_train_scaled, y_train_encoded)

    # Below makes a prediction:
    y_pred = model.predict(X_test_scaled)

    # Below computes metrics for evaluation:
    # Below is the % of correct predictions.
    acc = accuracy_score(y_test_encoded, y_pred)
    # Below is how pure each predicted class is.
    precision = precision_score(y_test_encoded, y_pred, average = 'macro')
    # Below is how well each class is recovered.
    recall = recall_score(y_test_encoded, y_pred, average = 'macro')
    # Below is a balanced metric even for imbalanced classes.
    mcc = matthews_corrcoef(y_test_encoded, y_pred)

    # Below logs the metric by storing the model's performance inside the experiment tracker.
    exp.log_metric("accuracy", acc)
    exp.log_metric("precision", precision)
    exp.log_metric("recall", recall)
    exp.log_metric("mcc", mcc)

    # Below displays the results with streamlit
    streamlit.write("📊 Model Performance:")
    streamlit.write(f"- Accuracy: `{acc:.4f}`")
    streamlit.write(f"- Precision: `{precision:.4f}`")
    streamlit.write(f"- Recall: `{recall:.4f}`")
    streamlit.write(f"- MCC: `{mcc:.4f}`")
```

Section 4 - Part 4: Hyperparameter Tuning:

```
# Now we are going to analyze the results from our 
# random Forest hyperparameter tuning epxeriments. 

# Below imports all that is needed:
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, matthews_corrcoef
from datetime import datetime # Used to generate unique run names.
import itertools # Used to generate all hyperparameter combinations.

# Below defines the hyperparameter grid, which is the set of value you want to try for each hyperparameter.
param_grid = {
    # Below says we try 100 and 200 trees.
    "n_estimators": [100, 200],
    # Below says we try depth of 3 and depth of 5.
    "max_depth": [3, 5],
    # Below says we try leaf sizes of 1,2 and 5.
    "min_samples_leaf": [1, 2, 5],
    # Below says we try feature selection strategies "sqrt" and "log2"
    "max_features": ['sqrt', 'log2']
}

# Below initializes the list to store results
results = []

# Below generates all combinations of parameters
# itertools.product creates every possible combiantion. 
# zip pairs parameter names with values. 
# dict() turns each pair into a dictionary.
# In total we train 24 different random forest models. We train so many as we dont know 
# ahead of time which combiantion of hyperparameters will perform the best. 
param_combinations = [dict(zip(param_grid.keys(), v)) for v in itertools.product(*param_grid.values())]

# Below loops through every possible parameter combiantoin:
for params in param_combinations:
    # Add a fixed random state to params for reproducibility across runs.
    params['random_state'] = 42
    
    # Below creates a unique run name with timestamp and params summary
    run_name = f"RF_tune_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

    # Below creates an experiment run, and everything inside of this block gets logged to Snowflake.
    # This includes: Hyperparameters, metrics, run name.
    with exp.start_run(run_name):
        # Below logs the hyperparameters that are used in this run.
        exp.log_params(params)

        # Below trains a new random forest model with the current hyperparamters.
        model = RandomForestClassifier(**params)
        model.fit(X_train_scaled, y_train_encoded)

        # Below pedicts the test set.
        y_pred = model.predict(X_test_scaled)

        # Below calculates the evaluation metrics
        acc = accuracy_score(y_test_encoded, y_pred) # Overall correctness
        precision = precision_score(y_test_encoded, y_pred, average = 'macro') # How pure predictions are across classes 
        recall = recall_score(y_test_encoded, y_pred, average = 'macro') # How well each class is recovered.
        mcc = matthews_corrcoef(y_test_encoded, y_pred) 

        # Log metrics to snowflake
        exp.log_metric("accuracy", acc)
        exp.log_metric("precision", precision)
        exp.log_metric("recall", recall)
        exp.log_metric("mcc", mcc)

        # Store results in snowflake by creating a python list of dictionaries. 
        # This is done so we can convert to a dataframe, sort by accuracy, visualize results.
        results.append({
            'run_name': run_name,
            'n_estimators': params['n_estimators'],
            'max_depth': params['max_depth'],
            'min_samples_leaf': params['min_samples_leaf'],
            'max_features': params['max_features'],
            'accuracy': acc,
            'precision': precision,
            'recall': recall,
            'mcc': mcc
        })

        streamlit.write(f"Parameters: {params}")
```

```
# With hyperparameter tuning finished, we can analyze the results of the 24 models 
# to find the best performing one.

# Below creates a DataFrame from the results list by converting it into a dataframe. 
results_df = panda.DataFrame(results)

# Below displays the summary statistics table in streamlit.
# This ensures the entire tuning table is displayed, the best value in each metric column is highlighted in green, 
# and this is done since it is easy to visually scan for top performers. 
streamlit.write("📊 Model Performance Summary:")
streamlit.dataframe(results_df.style.highlight_max(subset = ['accuracy', 'precision', 'recall', 'mcc'], color = "green"))

# Below displays the best performing configuration
# idxmax() finds the row index with thehighest accruacy.
# .loc[] retrives that entire row
# Thus best_model is a single row containing the information on:
# the best hyperparameters, the best accruacy, the best precision, the best recall, 
# the best MCC, the run name.
best_model = results_df.loc[results_df['accuracy'].idxmax()]


# Below displays the best model's performance.
streamlit.subheader("🏆 Best Model Configuration:")
streamlit.write("Learning Algorithm: `Random Forest`")
streamlit.write(f"Accuracy: `{best_model['accuracy']:.4f}`")
streamlit.write(f"Precision: `{best_model['precision']:.4f}`")
streamlit.write(f"Recall: `{best_model['recall']:.4f}`")
streamlit.write(f"MCC: `{best_model['mcc']:.4f}`")
streamlit.write(f"Learning Parameters:")
streamlit.write(f"- n_estimators: `{best_model['n_estimators']}`")
streamlit.write(f"- max_depth: `{best_model['max_depth']}`")
streamlit.write(f"- min_samples_leaf: `{best_model['min_samples_leaf']}`")
streamlit.write(f"- max_features: `{best_model['max_features']}`")
```

Section 4 - Part 5: Model Registry

```
# Now we register the best model to Snowflake's Model 
# Registry for deployment and management 

# We now will train a final model with the best models' paramaters
# We use these optimal parameters: 
# Number of estimators: 100
# Maximum depth: 5
# Minimum samples per leaf: 2
# Maximum features: sqrt


# Below extracts the configuration from best models' previous results
# Best_model is a row fro the results dataframe, and we pull out the best values for each  
# hyperparameter, and then convert them to the correct types. 
best_params = {
    'n_estimators': int(best_model['n_estimators']),
    'max_depth': int(best_model['max_depth']),
    'min_samples_leaf': int(best_model['min_samples_leaf']),
    'max_features': best_model['max_features'],
    'random_state': 42 # This is done for reproducibility.
}

# Below trains the final and best model:
# Below creates the model, by building a random forest using the optimal settgins. 
final_model = RandomForestClassifier(**best_params)
# Below traisn the model using the scaled training data.
final_model.fit(X_train_scaled, y_train_encoded)

# This final models is now:
# evaluated, registered, deployed, and used for predictions. 
```

```
# Now we register the best performing model in snowflake's model registry. 

# Below creates a model registry instance
# Registry is snowflakes system for storing, versioning, and managing ML models.
from snowflake.ml.registry import Registry
# Datetime is used to generate a unqiue version name.
from datetime import datetime


# Below temporarily suppresses warnings
warnings.filterwarnings('ignore', category = RuntimeWarning)

# Below connects the registry to your active snowflake session. 
registry = Registry(sesh)

# Below cleans the column names by replacing spaces with underscores
# This is done based on the snowflake model resgitries' preferences for no spaces, 
# no special characters, and consistent naming. 
X_train_clean = X_train_scaled.copy()
X_train_clean.columns = X_train_clean.columns.str.replace(' ', '_')

# Below defines the model name and version.
model_name = "BEAR_SPECIES_CLASSIFIER"
model_version = f"v_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

# Below registers the model in snowflake. 
# We store the following in the resgitry: 
# The trained model (final_model), model name and version, sample input data, performance metrics, 
# options (case_sensitive=True, and min_positive_value=1e-10 to avoid log(0) issues), and a comment.   
model_ref = registry.log_model(
    model = final_model,
    model_name = model_name,
    version_name = model_version,
    sample_input_data = X_train_clean.head(5),
    metrics = {
        'accuracy': float(best_model['accuracy']),
        'precision': float(best_model['precision']),
        'recall': float(best_model['recall']),
        'mcc': float(best_model['mcc'])
    },
    options = {
        "case_sensitive": True,
        "min_positive_value": 1e-10  # Add small constant to prevent log(0)
    },
    comment = "Best performing Random Forest model from hyperparameter tuning"
)

# Below resets warnings to default
warnings.resetwarnings()

# Below displays the results with streamlit:
streamlit.write("✅ Model successfully registered!")
streamlit.write(f"Model Name: `{model_name}`")
streamlit.write(f"Version: `{model_version}`")
```
